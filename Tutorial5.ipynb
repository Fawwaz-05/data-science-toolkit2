```python
import pandas as pd


file = '/content/drive/MyDrive/salary.csv'
df = pd.read_csv(file)

display(df.head())
```

```python
print(df.columns[6])
```

```python
df.rename(columns={df.columns[6] : 'Salary'}, inplace = True)
display(df.head())
```

```python
#Convert Format for Columns salary so there is no 56,000 to 56000



df['Salary']= df['Salary'].replace(','',',regex=True)
df['Salary']= df['Salary'].astype(int)
df.head()
```

```python
df.info() #info find missing value
```

```python
#Rename the columns

df.rename(columns={df.columns[7] : 'Compensation'}, inplace = True)
display(df.head())
```

```python
#count missing value in columns


print("Sum of null  :" , df['Compensation'].isna().sum())
```

```python
#Delete The columns


df.rename(columns={df.columns[5] : 'Context'}, inplace = True)
df.rename(columns={df.columns[9] : 'Other currency'}, inplace = True)
df.rename(columns={df.columns[10] : 'Salary Context'}, inplace = True)


#Delete columns

df = df.drop(
    columns = ['Other currency' , 'Other currency','Salary Context']
             )

df.head()




```

```python
#delete Row have mising value

df_clean = df.dropna()


#see the delete row
 # ~ for find index NOT in df_clean


droprow = df[~df.index.isin(df_clean.index)]




print("DELETED")
print(droprow)
```

```python
#KEEP row atleast 2 non mising value
# (thresh = " ")

df.dropna(thresh=2)

#delete went all is missing

df.dropna(how = 'all')

#compare number of row before and after

```

```python
#Imputation ( fill the mising value )

#  mean , median , most frequent


from sklearn.impute import SimpleImputer

#look and find most kerap value
imputer = SimpleImputer(strategy='most_frequent')

# Create a proper copy of df
df_imputed = df.copy()

# Iterate over the dataframe columns
for column in df_imputed.columns:
  if df_imputed[column].isna().any():
    # SimpleImputer expects 2D array, so we use [[column]]
    df_imputed[[column]] = imputer.fit_transform(
        df_imputed[[column]]
    )

# Print sum of missing values after imputation
print(df_imputed.isna().sum())

df_imputed.head()
```
