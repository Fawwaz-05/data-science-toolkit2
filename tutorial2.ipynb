```python
import numpy as np


array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is numpy 2x2 array")
print(array_2x2)
```

```python
random3x4 = np.random.rand(3,4)

print("random 3x4 array")
print(random3x4)
```

```python
ones_array = np.ones((3,3))

zeros_array = np.zeros((3,3))

print("array one")
print(ones_array)
print("array zero")
print(zeros_array)
```

```python
random6x3 = np.random.rand(6,3)

print("random 6x3 array")

print(np.transpose(random6x3))
```

```python
random3x4 = np.random.rand(3,4)
random4x5 = np.random.rand(4,5)

dot = random3x4 @ random4x5

print(dot)
```
